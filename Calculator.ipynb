```python
x= "##### Calculator #######"
print(x.center(130))    # String method

y= input("Enter symbols- '+','-','*','/','^','%(Remainder)'")

if y=="+":  # Addition
    a= float(input("Enter First value:"))
    b= float(input("Enter second value:"))
    i=a+b
    print("The Addition of these two numbers is:",i)
elif y=="-":  #Subtraction"
    a= float(input(" Enter First value:"))
    b= float(input(" Enter second value:"))
    i=a+b
    print("The Subtraction of these two numbers is:",i)
elif y=="*":  #Multiplication
    a= float(input(" Enter First value:"))
    b= float(input(" Enter second value:"))
    i=a*b
    print("The Multiplication of these two numbers is:",i)
elif y=="/":  # Division
    a= float(input(" Enter First value:"))
    b= float(input(" Enter second value:"))
    
    if b!=0:
        i=a/b
        print(i)
    else:
        print("Invalid input: Zero can not be divided by any no.")
elif y=="%":
    a= int(input(" Enter First value:"))
    b= int(input(" Enter second value:"))
    i=a%b
    print(i)
elif y=="^":
    a= int(input(" Enter First value:"))
    b= int(input(" Enter second value:"))
    i=a**b
    print(i)
else:
    print("Invalid operator")
```
