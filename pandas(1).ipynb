```python
import pandas as pd
info={"one":pd.Series([1,2,3,4,5],index=["a","b","c","d","e"]),"two":pd.Series([1,2,3,4,5,6],index=["a","b","c","d","e","f"])}
f=pd.DataFrame(info)
print("Add new columns by passing series")
f["three"]=pd.Series([20,40,60],index=["a","b","c"])
print(f)
print("Add new columns using existing DataFrame columns")
f["four"]=f["one"]+f["three"]
print(f)
```

```python
import pandas as pd
info={"one":pd.Series([1,2],index=["a","b"]),"two":pd.Series([1,2,3],index=["a","b","c"])}
f=pd.DataFrame(info)
print("The DataFrame:")
print(f)

#By del function:

print("Delete the first column:")
del f["two"]
print(f)

#By pop function:
print("Delete another column:")
f.pop("one")
print(f)
```
